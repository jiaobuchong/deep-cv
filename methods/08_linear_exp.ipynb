# Backbone and test-resolution checks (`08_linear_exp`)

Two cheap checks that train no backbone. They test two possible reasons for the gap to the top of the Phase 1 leaderboard (84.07% vs our 76.22%). **The results are for the report only, as a post-deadline analysis.** Nothing here is submitted: Phase 2 is closed, and test predictions must come from the Phase 1 model.

> For Phase 2 test-set predictions, use only the `PHASE2_TEST` switch in the main notebook. Do not use this notebook.

**Direction 1: another pretrained version of ViT-B/16.** Our backbone, `augreg2_in21k_ft_in1k`, was pretrained on ImageNet-21k and then fine-tuned on the 1,000 classes of ImageNet-1k. The project brief also names `vit-base-patch16-224-in21k`, a version that stops after ImageNet-21k. Its features have not been narrowed to 1,000 classes, so they may suit our 10,000 fine-grained classes better. We measure zero-shot and linear-probe accuracy for two such checkpoints, `augreg_in21k` and `orig_in21k`. A frozen backbone makes this a direct test of feature quality. Our backbone, from batch `1102_1149`: zero-shot **43.95%**, linear probe **61.94%**.

**Direction 2: a larger test image size.** We load a trained run, `1019_1120_1_full` by default (full fine-tuning, prototype head, 3 epochs, trained at 224 px, holdout 73.38%), and evaluate it on the same 20,000 holdout images at larger input sizes. ViT accepts other sizes by interpolating its position embeddings, with no training. RandomResizedCrop makes objects look larger during training than in test images, so a slightly larger test size often helps (FixRes, Touvron et al., 2019). The 224 px row must match the run's own holdout score; this checks that the holdout split is the one the run was trained with.

**How to run**: Run all cells on an A100. `RUN_BACKBONES` and `RUN_RESOLUTION` in §1 turn each part off. It takes about 30 minutes and 0.3 GPU-hours. Results: `cs604/runs/<batch>_<i>_<backbone>_<head>/` for direction 1, and `cs604/batches/<batch>/` (`results.md`, `resolution.md`) for both.

**How to read the results**:
- Direction 1: a linear probe clearly above 61.94% (say 2 points or more) means the backbone version matters. The next step would be one 3-epoch full fine-tuning run with that backbone.
- Direction 2: if accuracy rises above 224 px, resolution matters, and training the last epoch at the larger size would likely help more. If it falls, the model needs training at that size before we can tell.

## 3. Prepare the data

- In each new session, the data is first copied to the Colab local disk (`DATA_DIR`), and the images are then read from there. Reading 450,000 small images one by one from the internet or from Drive is very slow.
- By default the data comes from Hugging Face (`DATA_FROM_HF = True`). On Colab this runs at about 495 MB/s, so the 40 GB `train.tar.gz` downloads in just over a minute. The archive is deleted after it is unpacked. During download and unpacking, the local disk needs about 80 GB of free space.
- With `DATA_FROM_HF = False`, the backup on Drive is used instead (`DRIVE_DATA_DIR`; save it there first with `cs604_download_to_drive.ipynb`). It is unpacked directly from Drive at about 33 MB/s, which takes about 20 minutes.
- Images keep their original size (longest side 500 pixels) and are not resized in advance: on an A100 runtime, the CPU decodes fast enough to keep up with the GPU.
- When the data is ready, a marker file is written, so this step is skipped if you run it again in the same session.
- Only the training set is used here. The official val set has no labels, and the submission file is only written in the main notebook.

```python
def prepare_split(split):
    """Put {split}.txt and the images of {split}.tar.gz on the local Colab disk under DATA_DIR. The files come from Hugging Face,
    or, with DATA_FROM_HF = False, from the copy on Drive. Returns the path of the local {split}.txt"""
    list_path = os.path.join(DATA_DIR, f"{split}.txt")
    marker = os.path.join(DATA_DIR, f".{split}_ready")
    if os.path.exists(marker):  # already done in this session
        return list_path
    t0 = time.time()
    if DATA_FROM_HF:
        hf_hub_download(HF_REPO, f"{split}.txt", repo_type="dataset", local_dir=DATA_DIR)
        tar_path = hf_hub_download(HF_REPO, f"{split}.tar.gz", repo_type="dataset", local_dir=os.path.join(DATA_DIR, "hf"))
        log(f"[{split}] downloaded from Hugging Face in {(time.time() - t0) / 60:.1f} min")
    else:
        shutil.copy(os.path.join(DRIVE_DATA_DIR, f"{split}.txt"), list_path)
        tar_path = os.path.join(DRIVE_DATA_DIR, f"{split}.tar.gz")
    log(f"[{split}] extracting {tar_path} ...")
    subprocess.run(["tar", "-xzf", tar_path, "-C", DATA_DIR], check=True)
    if DATA_FROM_HF:
        os.remove(tar_path)  # the images are out; free the disk space the archive takes (40 GB for train)
    with open(list_path) as f:
        n_expected = sum(1 for line in f if line.strip())
    n_found = sum(len(files) for _, _, files in os.walk(os.path.join(DATA_DIR, split)))
    assert n_found == n_expected, f"{split}: found {n_found} images, expected {n_expected}"
    open(marker, "w").close()
    log(f"[{split}] {n_found} images ready in {(time.time() - t0) / 60:.1f} min")
    return list_path


def read_list(path):
    """Each line is '<image path> <class id>', e.g. 'train/class_0000/9aef28bf-....jpg 0'"""
    items = []
    with open(path) as f:
        for line in f:
            parts = line.split()
            if parts:
                items.append((parts[0], int(parts[1])))
    return items


TRAIN_ALL = read_list(prepare_split("train"))
log(f"train {len(TRAIN_ALL):,} images | {NUM_WORKERS} loader workers")
```

## 4. Model (zero-shot / linear probe)

- `apply_method` makes all parameters, including the head, not trainable, so this notebook has no training loop.
- The head weights come entirely from `init_head` in §6.

```python
def is_head(name):
    return name in ("head.weight", "head.bias")


def resolve_config(overrides):
    """Defaults from section 1 + one experiment's overrides → the full config of this run"""
    unknown = set(overrides) - set(RUN_KEYS)
    assert not unknown, f"unknown keys in experiment config: {sorted(unknown)}"
    cfg = SimpleNamespace(**{**BASE_CONFIG, **overrides})
    assert cfg.HEAD_INIT in ("proto", "lp"), f"HEAD_INIT must be 'proto' or 'lp', not {cfg.HEAD_INIT!r}"
    assert cfg.HOLDOUT_PER_CLASS > 0, "this notebook evaluates on the holdout set, so keep at least 1 image per class"
    if cfg.BATCH_SIZE is None:
        cfg.BATCH_SIZE = 256 if VRAM_GB >= 40 else 128 if VRAM_GB >= 24 else 64
    return cfg


def create_model(cfg):
    """Pretrained ViT-B/16 from timm; its 1000-class ImageNet head is replaced by a new, randomly initialized 10,000-class head"""
    return timm.create_model(cfg.MODEL_NAME, pretrained=True, num_classes=cfg.NUM_CLASSES).to(DEVICE)


def apply_method(net, cfg):
    """Zero-shot / linear probe: nothing is trained with gradients; the head was already set in section 6"""
    return net.requires_grad_(False)


def param_report(net, cfg, ctx):
    """Trainable parameter counts. The backbone is frozen; a linear-probe head was trained in section 6, so it counts as trainable"""
    n_train = sum(p.numel() for p in net.parameters() if p.requires_grad)  # 0: apply_method froze everything
    lp = cfg.HEAD_INIT == "lp"
    return {"backbone_params": ctx.n_backbone, "head_params": ctx.n_head,
            "head_mode": "linear probe" if lp else "class prototypes (not trained)",
            "trainable_excl_head": n_train, "trainable_pct": 100 * n_train / ctx.n_backbone,
            "trainable_incl_head": n_train + (ctx.n_head if lp else 0)}
```

## 5. Dataset and local holdout set

The official val set has no labels, so we keep `HOLDOUT_PER_CLASS` images per class from the training set as a local holdout set. It is only used to measure accuracy, never for training. The split is fixed by `SEED`, so it is the same in every experiment. This notebook does not train the backbone, so images are only resized and center-cropped, with no data augmentation.

```python
def split_holdout(items, k, seed):
    """Hold out k images of every class: shuffle each class's image list with a fixed seed and take the first k"""
    by_class = {}
    for path, label in items:
        by_class.setdefault(label, []).append(path)
    rng = random.Random(seed)
    train, holdout = [], []
    for label in sorted(by_class):
        paths = sorted(by_class[label])
        rng.shuffle(paths)
        holdout += [(p, label) for p in paths[:k]]
        train += [(p, label) for p in paths[k:]]
    return train, holdout


class ImageList(Dataset):
    def __init__(self, items, transform):
        self.items, self.transform = items, transform

    def __len__(self):
        return len(self.items)

    def __getitem__(self, i):
        path, label = self.items[i]
        with Image.open(os.path.join(DATA_DIR, path)) as im:
            img = im.convert("RGB")
        return self.transform(img), label


BICUBIC = T.InterpolationMode.BICUBIC


def make_ctx(model, cfg):
    """Data setup for this run: train/holdout split, image transform, normalization values, param counts"""
    data_cfg = timm.data.resolve_model_data_config(model)
    size = data_cfg["input_size"][-1]
    train_items, holdout_items = split_holdout(TRAIN_ALL, cfg.HOLDOUT_PER_CLASS, cfg.SEED)
    return SimpleNamespace(
        img_size=size, train_items=train_items, holdout_items=holdout_items,
        mean=torch.tensor(data_cfg["mean"], device=DEVICE).view(1, 3, 1, 1) * 255,
        std=torch.tensor(data_cfg["std"], device=DEVICE).view(1, 3, 1, 1) * 255,
        # Only resize/crop on the CPU and output uint8; normalize on the GPU to save CPU work
        eval_tf=T.Compose([T.Resize(int(size / data_cfg["crop_pct"]), interpolation=BICUBIC),
                           T.CenterCrop(size), T.PILToTensor()]),
        n_backbone=sum(p.numel() for n, p in model.named_parameters() if not is_head(n)),
        n_head=sum(p.numel() for p in model.head.parameters()),
    )


class Loader(DataLoader):
    """Before starting new worker processes, run garbage collection in the main process. Otherwise an old, interrupted DataLoader is copied into the new workers,
    and when it is collected there it prints lots of harmless AssertionError: can only test a child process messages."""

    def __iter__(self):
        gc.collect()
        return super().__iter__()


def make_loader(items, ctx, cfg):
    """Center crops, in order, 2 × BATCH_SIZE images per batch (no gradients, so a big batch fits)"""
    return Loader(ImageList(items, ctx.eval_tf), batch_size=2 * cfg.BATCH_SIZE, num_workers=NUM_WORKERS, pin_memory=True)


def to_input(x, ctx):
    """uint8 images (B, 3, 224, 224) → normalized float32 on the GPU"""
    return (x.to(DEVICE, non_blocking=True).float() - ctx.mean) / ctx.std


def autocast():
    """bf16 mixed precision: matrix multiplies run in bfloat16, the weights stay float32"""
    return torch.autocast("cuda", dtype=torch.bfloat16)


def free_gpu():
    """Free the GPU memory left by the previous run and reset the peak memory stats"""
    gc.collect()
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats()
```

## 6. Classifier head

- `proto` (zero-shot): use the frozen backbone to extract features for all training images. For each class, the average direction of its features becomes the weight of that class (the class prototype), multiplied by a temperature. No parameters are trained.
- `lp` (linear probe): start from the class prototypes, then train the head for `LP_EPOCHS` epochs on features computed in advance. Only the head is trained, so it is fast.

The second experiment reuses the features from the first one (`FEATURE_CACHE`), but the feature extraction time still counts toward its GPU time, so the time recorded for both experiments is accurate.

```python
@torch.inference_mode()
def extract_features(net, items, ctx, cfg):
    """Features of the frozen backbone: the class-token output after the final LayerNorm, 768 numbers per image"""
    net.eval()
    feats, labels = [], []
    for x, y in tqdm(make_loader(items, ctx, cfg), desc="train features", leave=False):
        with autocast():
            f = net.forward_head(net.forward_features(to_input(x, ctx)), pre_logits=True)
        feats.append(f.half())
        labels.append(y.to(DEVICE))
    return torch.cat(feats), torch.cat(labels)


@torch.inference_mode()
def class_prototypes(feats, labels, num_classes):
    """Mean direction of each class: L2-normalize every feature, add them up per class, normalize the sums"""
    protos = torch.zeros(num_classes, feats.shape[1], device=feats.device)
    for i in range(0, len(feats), 65536):
        protos.index_add_(0, labels[i:i + 65536], F.normalize(feats[i:i + 65536].float(), dim=1))
    return F.normalize(protos, dim=1)


@torch.inference_mode()
def fit_logit_scale(protos, feats, labels, n=10000):
    """Pick a good temperature for cosine similarity, so the softmax of the prototype head is not too flat or too sharp"""
    idx = torch.randperm(len(feats), device=feats.device)[:n]
    cos = F.normalize(feats[idx].float(), dim=1) @ protos.T
    losses = {s: F.cross_entropy(cos * s, labels[idx]).item() for s in (5, 10, 15, 20, 30, 40, 60, 80)}
    return min(losses, key=losses.get)


def linear_probe(feats, labels, weight, bias, cfg):
    """Train only the 768 → 10,000 head on the cached features, starting from the prototype head"""
    head = nn.Linear(feats.shape[1], cfg.NUM_CLASSES).to(DEVICE)
    with torch.no_grad():
        head.weight.copy_(weight)
        head.bias.copy_(bias)
    opt = torch.optim.AdamW(head.parameters(), lr=cfg.LP_LR, weight_decay=cfg.LP_WD)
    total = cfg.LP_EPOCHS * math.ceil(len(feats) / cfg.LP_BATCH)
    sched = torch.optim.lr_scheduler.LambdaLR(opt, lambda s: 0.5 * (1 + math.cos(math.pi * s / total)))
    for _ in tqdm(range(cfg.LP_EPOCHS), desc="linear probe", leave=False):
        perm = torch.randperm(len(feats), device=feats.device)
        for i in range(0, len(feats), cfg.LP_BATCH):
            idx = perm[i:i + cfg.LP_BATCH]
            loss = F.cross_entropy(head(feats[idx].float()), labels[idx], label_smoothing=cfg.LABEL_SMOOTHING)
            opt.zero_grad(set_to_none=True)
            loss.backward()
            opt.step()
            sched.step()
    return head.weight.detach(), head.bias.detach()


FEATURE_CACHE = {}  # the features extracted by the first run, reused by the second run


def cached_features(model, ctx, cfg):
    """Frozen-backbone features of the training images. Extracted once; later runs with the same model and split reuse them"""
    key = (cfg.MODEL_NAME, cfg.HOLDOUT_PER_CLASS, cfg.SEED)
    if key in FEATURE_CACHE:
        log("Reusing the backbone features extracted by the previous run")
        return FEATURE_CACHE[key]
    t0 = time.time()
    feats, labels = extract_features(model, ctx.train_items, ctx, cfg)
    torch.cuda.synchronize()
    # Keep them on the CPU between runs, so they don't count toward the next run's GPU memory
    FEATURE_CACHE[key] = dict(feats=feats.cpu(), labels=labels.cpu(), seconds=time.time() - t0)
    log(f"Extracted features for {len(feats):,} train images in {FEATURE_CACHE[key]['seconds'] / 60:.1f} min")
    return FEATURE_CACHE[key]


def init_head(model, ctx, cfg):
    """Set the head from the class prototypes (zero-shot), or from a linear probe trained from them (lp).
    Returns the seconds spent; feature extraction counts too, even when the previous run already extracted the features"""
    cached = cached_features(model, ctx, cfg)
    t0 = time.time()
    feats, labels = cached["feats"].to(DEVICE), cached["labels"].to(DEVICE)
    protos = class_prototypes(feats, labels, cfg.NUM_CLASSES)
    scale = fit_logit_scale(protos, feats, labels)
    # The head works on unnormalized features: divide by the mean feature norm so logits ≈ scale × cosine similarity
    weight = protos * (scale / feats.float().norm(dim=1).mean())
    bias = torch.zeros(cfg.NUM_CLASSES, device=DEVICE)
    if cfg.HEAD_INIT == "lp":
        weight, bias = linear_probe(feats, labels, weight, bias, cfg)
    with torch.no_grad():
        model.head.weight.copy_(weight)
        model.head.bias.copy_(bias)
    torch.cuda.synchronize()
    seconds = cached["seconds"] + time.time() - t0
    source = "a linear probe started from the class prototypes" if cfg.HEAD_INIT == "lp" else "the class prototypes"
    log(f"Head set from {source} (prototype logit scale {scale}) in {seconds / 60:.1f} min incl. feature extraction, "
        f"counted in this run's GPU time")
    return seconds
```

## 7. Evaluation, inference latency and saving the head

- `evaluate` computes top-1 and top-5 on the holdout images. The images go through the backbone again instead of using the cached features, so this checks the whole pipeline from the start.
- Inference latency is measured with random inputs. It only counts the forward pass of the model (not image loading or decoding), and uses bf16. Batch size 1 is run 100 times and averaged; batch size 256 is run 20 times to get the throughput.
- `save_state` only saves the head, because no other parameters change.

```python
@torch.inference_mode()
def evaluate(net, loader, ctx):
    """Top-1 and top-5 accuracy on a labeled set"""
    net.eval()
    c1 = c5 = n = 0
    for x, y in loader:
        with autocast():
            top5 = net(to_input(x, ctx)).topk(5, dim=1).indices
        y = y.to(DEVICE)
        c1 += (top5[:, 0] == y).sum().item()
        c5 += (top5 == y[:, None]).any(dim=1).sum().item()
        n += len(y)
    return c1 / n, c5 / n


def save_state(net, path, **extra):
    """Save the classifier head, about 31 MB; nothing else has changed"""
    state = {n: p.detach().cpu() for n, p in net.named_parameters() if p.requires_grad or is_head(n)}
    torch.save({"state": state, **extra}, path)


@torch.inference_mode()
def measure_latency(net, batch_size, iters, ctx, warmup=10):
    """Average seconds per forward pass on random input: model compute only, no image reading or decoding"""
    net.eval()
    x = torch.randn(batch_size, 3, ctx.img_size, ctx.img_size, device=DEVICE)
    for _ in range(warmup):
        with autocast():
            net(x)
    torch.cuda.synchronize()
    t0 = time.perf_counter()
    for _ in range(iters):
        with autocast():
            net(x)
    torch.cuda.synchronize()
    return (time.perf_counter() - t0) / iters


def latency_report(net, ctx):
    lat_1 = measure_latency(net, 1, 100, ctx)
    lat_b = measure_latency(net, 256, 20, ctx)
    setting = f"{GPU_NAME}, bfloat16, no flip TTA"
    log(f"Inference latency (batch 1): {lat_1 * 1e3:.2f} ms/img | throughput (batch 256): {256 / lat_b:.0f} img/s | {setting}")
    return dict(latency_ms_bs1=lat_1 * 1e3, throughput_img_s=256 / lat_b, latency_setting=setting)
```

## 8. Full pipeline for one experiment

`run_experiment` runs these steps in order: build the model → set the head → freeze all parameters → save the head → evaluate on the holdout set → measure inference latency → write `summary.json`. Results are on Drive in `cs604/runs/<run_name>/`: `train_log.txt`, `config.json`, `summary.json` and `final.pt`.

```python
def run_experiment(overrides, run_name):
    """Run one experiment and return its summary. Everything is saved to cs604/runs/<run_name>/ on Drive"""
    global LOG_FILE
    cfg = resolve_config(overrides)
    run_dir = os.path.join(OUT_ROOT, "runs", run_name)
    os.makedirs(run_dir, exist_ok=True)
    LOG_FILE = os.path.join(run_dir, "train_log.txt")
    try:
        with open(os.path.join(run_dir, "config.json"), "w") as f:
            json.dump({**vars(cfg), "RUN_NAME": run_name, "overrides": overrides}, f, indent=2)
        log(f"\n========== {run_name} ==========")
        log(f"{METHOD}: backbone {cfg.MODEL_NAME}, head {cfg.HEAD_INIT}, holdout {cfg.HOLDOUT_PER_CLASS}/class"
            + (f", linear probe {cfg.LP_EPOCHS} epochs (lr {cfg.LP_LR}, wd {cfg.LP_WD}, batch {cfg.LP_BATCH})"
               if cfg.HEAD_INIT == "lp" else ""))
        log(f"Environment: {GPU_NAME} ({VRAM_GB} GB), bfloat16, {os.cpu_count()} CPUs | {VERSIONS}")
        random.seed(cfg.SEED)
        np.random.seed(cfg.SEED)
        torch.manual_seed(cfg.SEED)
        free_gpu()

        model = create_model(cfg)
        ctx = make_ctx(model, cfg)
        seconds = init_head(model, ctx, cfg)
        model = apply_method(model, cfg)
        params = param_report(model, cfg, ctx)
        log(f"Trainable (excl. head): {params['trainable_excl_head']:,} | head {ctx.n_head:,} params ({params['head_mode']})")
        save_state(model, os.path.join(run_dir, "final.pt"), config=vars(cfg))
        log(f"GPU time: {seconds / 3600:.2f} GPU-hours (feature extraction + classifier head; the backbone is not trained)")
        # The project doc asks for VRAM in the training log; reserved is closer to what nvidia-smi shows
        log(f"Peak VRAM: {torch.cuda.max_memory_allocated() / 1e9:.1f} GB allocated, "
            f"{torch.cuda.max_memory_reserved() / 1e9:.1f} GB reserved by PyTorch (GPU has {VRAM_GB} GB)")

        top1, top5 = evaluate(model, make_loader(ctx.holdout_items, ctx, cfg), ctx)
        log(f"Final holdout: top-1 {top1:.2%}, top-5 {top5:.2%}")
        latency = latency_report(model, ctx)

        summary = dict(run=run_name, method=METHOD, head_init=cfg.HEAD_INIT, model=cfg.MODEL_NAME, overrides=overrides,
                       **params, holdout_top1=top1, holdout_top5=top5, **latency, gpu=GPU_NAME, vram_gb=VRAM_GB,
                       gpu_hours=seconds / 3600, peak_vram_gb=torch.cuda.max_memory_allocated() / 1e9,
                       peak_vram_reserved_gb=torch.cuda.max_memory_reserved() / 1e9, versions=VERSIONS)
        with open(os.path.join(run_dir, "summary.json"), "w") as f:
            json.dump(summary, f, indent=2)
        log(f"== {run_name} finished: holdout top-1 {top1:.2%} | {params['head_mode']} | {summary['gpu_hours']:.2f} GPU-h")
        return summary
    finally:
        LOG_FILE = None
```

## 9. Direction 1: run the backbone experiments

Each run is named `<batch>_<i>_<backbone>_<head>`, e.g. `1105_1400_2_in21k_lp`.

```python
BACKBONE_TAG = {"vit_base_patch16_224.augreg2_in21k_ft_in1k": "in1k", "vit_base_patch16_224.augreg_in21k": "in21k",
                "vit_base_patch16_224.orig_in21k": "orig21k"}  # short names for the run folders
RESULTS = []
if RUN_BACKBONES:
    for i, exp in enumerate(EXPERIMENTS, 1):
        name = exp.get("MODEL_NAME", MODEL_NAME)
        tag = BACKBONE_TAG.get(name, name.split(".")[-1])
        result = run_experiment(dict(exp), f"{BATCH_NAME}_{i}_{tag}_{exp.get('HEAD_INIT', HEAD_INIT)}")
        RESULTS.append(result)
        if result["head_init"] == "proto" and result["holdout_top1"] < 0.05:
            raise RuntimeError(f"Zero-shot holdout top-1 is only {result['holdout_top1']:.1%} for {name}: "
                               "the data pipeline or the checkpoint is probably broken, stopping. Please check the log.")
```

## 11. Direction 2: the same trained model at larger test sizes

- `load_trained` rebuilds a trained run with `dynamic_img_size=True`: timm then interpolates the 14 × 14 grid of position embeddings to the input size (for example 18 × 18 at 288 px). The weights do not change.
- Only runs trained with the same holdout split (2 images per class, the same seed) are accepted, so the holdout images were never seen in training. The final models trained on all 45 images per class are therefore not usable here.
- Every size uses the evaluation transform of training: resize to size / 0.9, then a center crop.
- The 224 px row must match the run's own holdout score (73.38% for `1019_1120_1_full`); a large difference stops the cell.
- Results: `cs604/batches/<batch>/resolution.md` and `resolution.json`.

```python
def load_trained(run_name):
    """A trained full or partial fine-tuning run from cs604/runs/, built to accept any input size"""
    run_dir = os.path.join(OUT_ROOT, "runs", run_name)
    with open(os.path.join(run_dir, "config.json")) as f:
        saved = json.load(f)
    summary = {}
    if os.path.exists(os.path.join(run_dir, "summary.json")):
        with open(os.path.join(run_dir, "summary.json")) as f:
            summary = json.load(f)
    method = saved.get("METHOD") or summary.get("method")
    assert method in ("full", "partial"), f"{run_name}: only full or partial fine-tuning runs are supported, not {method!r}"
    assert saved.get("HOLDOUT_PER_CLASS") == HOLDOUT_PER_CLASS and saved.get("SEED") == SEED, (
        f"{run_name} was trained with holdout {saved.get('HOLDOUT_PER_CLASS')}/class and seed {saved.get('SEED')}, "
        "so its holdout images are not held out here")
    # Frozen weights of a partial run come from the pretrained model; the trained ones from final.pt
    net = timm.create_model(saved["MODEL_NAME"], pretrained=True, num_classes=saved["NUM_CLASSES"], dynamic_img_size=True)
    ck = torch.load(os.path.join(run_dir, "final.pt"), map_location="cpu", weights_only=False)
    missing, unexpected = net.load_state_dict(ck["state"], strict=False)
    assert not unexpected, f"unexpected keys: {unexpected[:5]}"
    log(f"Loaded {run_name} ({method}, {saved['MODEL_NAME']}): {len(ck['state'])} trained tensors from final.pt, "
        f"{len(missing)} kept from the pretrained model")
    return net.to(DEVICE).eval(), summary.get("holdout_top1")


def eval_at(net, size, ctx, cfg, crop_pct):
    """Holdout top-1 and top-5 at one input size, with the evaluation transform of training"""
    tf = T.Compose([T.Resize(int(size / crop_pct), interpolation=BICUBIC), T.CenterCrop(size), T.PILToTensor()])
    loader = Loader(ImageList(ctx.holdout_items, tf), batch_size=RES_BATCH, num_workers=NUM_WORKERS, pin_memory=True)
    return evaluate(net, loader, ctx)


RES_RESULTS = []
if RUN_RESOLUTION:
    assert all(s % 16 == 0 for s in RES_SIZES), "every size must be a multiple of the 16-pixel patch"
    cfg = resolve_config({})
    for run_name in RES_RUNS:
        free_gpu()
        net, reported = load_trained(run_name)
        ctx = make_ctx(net, cfg)
        crop_pct = timm.data.resolve_model_data_config(net)["crop_pct"]
        for size in RES_SIZES:
            t0 = time.time()
            top1, top5 = eval_at(net, size, ctx, cfg, crop_pct)
            grid = size // 16
            RES_RESULTS.append(dict(run=run_name, size=size, grid=grid, tokens=grid * grid + 1, holdout_top1=top1,
                                    holdout_top5=top5, seconds=time.time() - t0))
            log(f"{run_name} at {size} px ({grid} × {grid} patches): holdout top-1 {top1:.2%}, top-5 {top5:.2%} "
                f"({time.time() - t0:.0f} s)")
            if size == 224 and reported is not None and abs(top1 - reported) > 0.005:
                raise RuntimeError(f"At 224 px {run_name} gets {top1:.2%}, but its own holdout score is {reported:.2%}: "
                                   "the holdout split is probably not the one it was trained with. Stopping.")
        del net
    base = {r["run"]: r["holdout_top1"] for r in RES_RESULTS if r["size"] == 224}
    header = "| Run | Test size | Tokens | Holdout top-1 | vs 224 px (points) | Holdout top-5 |\n|---|---|---|---|---|---|"
    rows = [f"| {r['run']} | {r['size']} px | {r['tokens']} | {r['holdout_top1']:.2%} | "
            f"{100 * (r['holdout_top1'] - base[r['run']]):+.2f} | {r['holdout_top5']:.2%} |" if r["run"] in base else
            f"| {r['run']} | {r['size']} px | {r['tokens']} | {r['holdout_top1']:.2%} | – | {r['holdout_top5']:.2%} |"
            for r in RES_RESULTS]
    res_table = header + "\n" + "\n".join(rows)
    with open(os.path.join(BATCH_DIR, "resolution.md"), "w") as f:
        f.write(res_table + "\n")
    with open(os.path.join(BATCH_DIR, "resolution.json"), "w") as f:
        json.dump(RES_RESULTS, f, indent=2)
    log("\n" + res_table)
    display(Markdown(res_table))
```
